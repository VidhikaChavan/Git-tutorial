```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\chava\Downloads\student_performance.csv")
print(df)
df.sample(10)
```

```python
df=df.rename(columns={"weekly_self_study_hours":"study_hours"})
```

```python
df=df[["student_id","attendance_percentage","class_participation","study_hours","total_score","grade"]]
```

```python
df.groupby("grade")["study_hours"].mean()
df.groupby("grade")["student_id"].count()
df.groupby("total_score")["attendance_percentage"].max()
df.groupby("grade")["attendance_percentage"].min()
df.groupby("grade")["study_hours"].std()
```

```python
df.groupby("total_score")["study_hours"].agg(["min","max","mean","count"])
```

```python
df.groupby("grade")["total_score"].agg(min_marks="min",max_marks="max")#only two columns allowed
```

```python
df.groupby("grade").agg({"study_hours":"mean","total_score":"mean"})
```

```python
df.groupby("grade").agg(avg_hours_of_study=("study_hours","mean"),avg_total_score=("total_score","mean"))
```
